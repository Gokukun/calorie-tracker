```python
import numpy as np
import pandas as pd
from sklearn import linear_model
import joblib
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("model calorie.xlsx")
df
```

```python
df.corr()
```

```python
df.groupby("duration")["calories_burned"].mean().plot(kind="bar",color="skyblue")
plt.xlabel("duration")
plt.ylabel("Calories")
plt.title("calories burned with time")
           
```

```python
df.groupby("heart_rate")["duration"].mean().plot(kind="bar",color="hotpink")
plt.xlabel("heart_rate")
plt.ylabel("duration")
plt.title("heart_rate vs duration")
```

```python
x = df[["duration","heart_rate"]]
y = df["calories_burned"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=20,test_size=0.3)
```

```python
model = linear_model.LinearRegression()
model.fit(x_train,y_train)
```

```python
model.predict([[52, 120]])
```

```python
joblib.dump(model,"meet_project_joblib")
```

```python
y_pred=model.predict(x_test)
```

```python
model.score(x_test,y_test)
```
